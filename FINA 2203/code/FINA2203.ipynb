```python
# Problem 7
cf_A = {
    0: -49,
    1: 24,
    2: 19,
    3: 19,
    4: 17
}

cf_B = {
    0: -102,
    1: 20,
    2: 39,
    3: 52,
    4: 58
}

def irr_calc(cash_flows, wacc, tol=1e-6):
    NPV = sum(cash_flows[i]/(1+wacc)**i for i in cash_flows.keys())
    if abs(NPV) < tol:
        return wacc
    dNPV = -sum(i * cash_flows[i]/(1+wacc)**(i+1) for i in cash_flows.keys())
    wacc_new = wacc - NPV/dNPV
    return irr_calc(cash_flows, wacc_new)
print(f'The IRR of Project A is {irr_calc(cf_A, 0.05)*100:.2f}%\nThe IRR of Project B is {irr_calc(cf_B, 0.05)*100:.2f}%')
wacc = 0.055
NPV_A = sum(cf_A[i]/(1+wacc)**i for i in cf_A.keys())
NPV_B = sum(cf_B[i]/(1+wacc)**i for i in cf_B.keys())
print(f'The NPV of project A is ${NPV_A:.2f} and of project B is ${NPV_B:.2f}')
```

```python
# Problem 8
cf_A = {
    0: -98,
    1: 22,
    2: 29,
    3: 42,
    4: 52
}

cf_B = {
    0: -98,
    1: 52,
    2: 42,
    3: 29,
    4: 20
}

wacc = 0.105
NPV_A = sum(cf_A[i]/(1+wacc)**i for i in cf_A.keys())
NPV_B = sum(cf_B[i]/(1+wacc)**i for i in cf_B.keys())
print(f'The NPV of project A is ${NPV_A:.2f} and of project B is ${NPV_B:.2f}')
print(f'The IRR of Project A is {irr_calc(cf_A, 0.05)*100:.2f}%\nThe IRR of Project B is {irr_calc(cf_B, 0.05)*100:.2f}%')

def indifferent_proj(wacc, cf_A, cf_B, tol=1e-6):
    NPV_A = sum(cf_A[i]/(1+wacc)**i for i in cf_A.keys())
    NPV_B = sum(cf_B[i]/(1+wacc)**i for i in cf_B.keys())
    diff = NPV_B - NPV_A
    if abs(diff) < tol:
        return wacc
    else:
        d_diff = -sum(i*cf_B[i]/(1+wacc)**(i+1) for i in cf_B.keys()) + sum(i*cf_A[i]/(1+wacc)**(i+1) for i in cf_A.keys())
        wacc_new = wacc - diff/d_diff
        return indifferent_proj(wacc_new, cf_A, cf_B)
print(f'We will be indifferent between the two projects when the cost of capital is {indifferent_proj(0.05, cf_A, cf_B)*100:.2f}%')
wacc = indifferent_proj(0.05, cf_A, cf_B)
NPV_A = sum(cf_A[i]/(1+wacc)**i for i in cf_A.keys())
NPV_B = sum(cf_B[i]/(1+wacc)**i for i in cf_B.keys())
print(f'The NPV of project A is ${NPV_A:.2f} and of project B is ${NPV_B:.2f}')
```
