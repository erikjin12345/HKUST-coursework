```python
revenue = 1_460_000
maintainence_cost_base = 86_140
g = 0.04
i = 0.05
# Revenues and costs occur at the end of the year
# We want to run the operations as long as it is profitable, ie the revenue is greater than the cost

#a) The present value of the revenues
T_even = math.log(revenue/maintainence_cost_base) / math.log(1+g) + 1 # we add 1 as the first revenue and cost occur after 1 year
T = math.floor(T_even)
print(f'The operation is profitable for {T} years')
PV_rev = (revenue/i) * (1-(1+i)**(-T))
print(f'The present value of the revenues is ${PV_rev}')

#b) The present value of the costs
PV_cost = (maintainence_cost_base/(i-g)) * (1-((1+g)/(1+i))**(T))
print(f'The present value of the costs is ${PV_cost}')

#c) The net present value
print(f'The difference in the present value of the revenue and cost is {PV_rev - PV_cost - 14_600_000}')
```

```python
cash = 950_000
T = 30
balance_or = 750_000
T_res = 18.5 # just made payment
i = 0.055 # the interest rate on the mortgage
m = 12
N = T * m
r_eff = i / m
# We want to calculate the net sale value, the original sale value less the downpayment on the mortgage
# The total downpayment is the present value of the remaining annuities
# Annuties needs to paid monthly
A = (balance_or * r_eff) / (1 - (1 + r_eff)**(-N))
print(f'The monthly annuitiy payments are ${A}')
N_rem = m * (T - T_res)
pv_annuity = (A/r_eff) * (1 - (1 + r_eff)**(-N_rem)) # The present value if the remaining annuities

net_cash = cash - pv_annuity
print(f'The net cash after the mortgage downpayment is ${net_cash}')
```

```python
r = 0.072
loan_payments = 682 # per month
m = 12
N = 36 # number of months left on the loan
cash = 1_200
r_eff = r / 12
# We want to calculate the reduction in the amount of time left if we pay off $1200 in addition to $682 next month
current_balance = (loan_payments / r_eff) * (1 - (1 + r_eff)**(-N))
print(f'The current balance of the bond is ${current_balance}')

# we want to calculate the time required to pay off (from the number of periods we decrease by) if we pay 1200 at time t=1
balance_1 = (current_balance - (cash + loan_payments))*(1+r_eff) # present value of the remaining balance
N_left = -math.log(1 - (balance_1 * r_eff / loan_payments)) / math.log(1+r_eff) + 1
print(f'The new time to pay off the loan will be {N_left:.1f} months')
print(f'The reduction in the number of months will be {(N - N_left):.1f} months')
```
