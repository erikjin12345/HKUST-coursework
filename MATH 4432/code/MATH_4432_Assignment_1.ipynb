```python
# (h) Further exploration, del 1: Elite, acceptance rate och korrelationer

# 1. Jämför elit- och icke-elitskolor: medelvärde per variabel
print("Medelvärde per grupp (Elite):")
display(college.groupby('Elite', observed=True).mean(numeric_only=True).T.round(1))

# 2. Median och medel för ett urval nyckelvariabler
key_cols = ['Outstate', 'Expend', 'Grad.Rate', 'S.F.Ratio', 'PhD', 'perc.alumni']
print("Median och medel för nyckelvariabler:")
display(college.groupby('Elite', observed=True)[key_cols].agg(['median', 'mean']).round(1))

# 3. Acceptance rate
college['AcceptRate'] = college['Accept'] / college['Apps']

print("Acceptance rate per Elite:")
display(college.groupby('Elite', observed=True)['AcceptRate'].describe().round(3))
print("Acceptance rate per Private:")
display(college.groupby('Private')['AcceptRate'].describe().round(3))
print("Antal skolor per Private x Elite:")
display(pd.crosstab(college['Private'], college['Elite']))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
college.boxplot(column='AcceptRate', by='Elite', ax=axes[0])
axes[0].set_title('Acceptance rate vs Elite')
college.boxplot(column='AcceptRate', by='Private', ax=axes[1])
axes[1].set_title('Acceptance rate vs Private')
axes[2].scatter(college['AcceptRate'], college['Grad.Rate'], alpha=0.5)
axes[2].set_xlabel('Acceptance rate')
axes[2].set_ylabel('Grad.Rate')
axes[2].set_title('Grad.Rate vs acceptance rate')
fig.suptitle('')
fig.tight_layout()
plt.show()

# 4. Korrelationer med Grad.Rate
corr_grad = college.corr(numeric_only=True)['Grad.Rate'].drop('Grad.Rate').sort_values()
print("Korrelation med Grad.Rate:")
display(corr_grad.round(3))

fig, ax = plt.subplots(figsize=(8, 5))
corr_grad.plot.barh(ax=ax)
ax.set_xlabel('Korrelation med Grad.Rate')
ax.axvline(0, color='black', linewidth=0.8)
fig.tight_layout()
plt.show()

# 5. Felaktiga värden: Grad.Rate över 100 %
print("Skolor med Grad.Rate > 100:")
display(college.loc[college['Grad.Rate'] > 100, ['Private', 'Apps', 'Grad.Rate']])
```

```python
# Exercise 8
auto = load_data('auto')
import statsmodels.api as sm
X = auto['horsepower']
Y = auto['mpg']
X = sm.add_constant(X)
lin_mod = sm.OLS(Y, X)
lin_reg = lin_mod.fit()
print(lin_reg.summary())
print(f'The predicted miles per gallon with 98 horsepowers: {-0.1578*98 + 39.9359}')
```

```python
from scipy.stats import t


def get_residual_se(y_true, y_pred, df):
    # RSS = Sum of squared residual errors
    RSS = ((y_true - y_pred) ** 2).sum()
    s = (RSS / df) ** 0.5
    return s


def get_pred_se(x, x0, CI=True):
    n = len(x)  # Rätt: len(x) istället för len(auto)
    x_mean = sum(x) / n
    s_xx = sum((x - x_mean) ** 2)

    if CI:
        inside = (1 / n) + ((x0 - x_mean) ** 2) / s_xx 
    else:
        inside = (
            1 + (1 / n) + ((x0 - x_mean) ** 2) / s_xx
        )  

    return inside**0.5


def get_CI(x0, ci_level, p, data, x_col, y_col, model):
    n = len(data)  
    df = n - p
    alpha = 1 - ci_level

    y_0_hat = model.predict([1.0, x0])[0]
    t_crit = float(t.ppf(1 - alpha / 2, df=df))

    s = get_residual_se(data[y_col], model.fittedvalues, df)
    scale_factor = get_pred_se(data[x_col], x0, CI=True)  

    margin = t_crit * s * scale_factor
    return [float(y_0_hat - margin), float(y_0_hat + margin)]


def get_PI(x0, ci_level, p, data, x_col, y_col, model):
    n = len(data)  
    df = n - p
    alpha = 1 - ci_level

    y_0_hat = model.predict([1.0, x0])[0]
    t_crit = float(t.ppf(1 - alpha / 2, df=df))

    s = get_residual_se(data[y_col], model.fittedvalues, df)
    scale_factor = get_pred_se(data[x_col], x0, CI=False)  

    margin = t_crit * s * scale_factor
    return [float(y_0_hat - margin), float(y_0_hat + margin)]


# Körning
x0 = 98
ci = get_CI(
    x0,
    ci_level=0.95,
    p=2,
    data=auto,
    x_col='horsepower',
    y_col='mpg',
    model=lin_reg,
)
pi = get_PI(
    x0,
    ci_level=0.95,
    p=2,
    data=auto,
    x_col='horsepower',
    y_col='mpg',
    model=lin_reg,
)

print(f'95% CI for x0={x0}: {ci}')
print(f'95% PI for x0={x0}: {pi}')
```

```python
def abline(ax, b, m, *args, **kwargs):
    "Add a line with slope m and intercept b to ax"
    xlim = ax.get_xlim()
    ylim = [m * xlim[0] + b, m * xlim[1] + b]
    ax.plot(xlim, ylim, *args, **kwargs)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(auto['horsepower'], auto['mpg'], alpha=0.5)
abline(ax, lin_reg.params['const'], lin_reg.params['horsepower'], 'r--', linewidth=2)
ax.set_xlabel('horsepower'); ax.set_ylabel('mpg')
plt.show()
```
