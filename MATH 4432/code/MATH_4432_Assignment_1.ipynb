```python
# Exercise 8
from ISLP import load_data
import pandas as pd
import matplotlib.pyplot as plt
college = pd.read_csv("College.csv")

college2 = pd.read_csv('College.csv', index_col=0) 
college3 = college.rename({'Unnamed: 0': 'College'},
axis=1)
college3 = college3.set_index('College')

college = college3
college

```

```python
# Statistics about the variables in the dataset
college.describe()
```

```python
# Scatterplots of three variables
columns = ["Top10perc", "Apps", "Enroll"]
pd.plotting.scatter_matrix(college[columns])
```

```python
#Boxplot
college.boxplot(column="Outstate", by="Private")
```

```python
college['Elite'] = pd.cut(college[ 'Top10perc'],
[0,50,100],
labels=['No', 'Yes'])
print(college["Elite"].value_counts())
college.boxplot(column="Outstate", by="Elite")
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

college['Apps'].plot.hist(bins=20, ax=axes[0, 0], title='Apps, 20 bins')
college['Outstate'].plot.hist(bins=10, ax=axes[0, 1], title='Outstate, 10 bins')
college['PhD'].plot.hist(bins=30, ax=axes[1, 0], title='PhD, 30 bins')
college['Grad.Rate'].plot.hist(bins=15, ax=axes[1, 1], title='Grad.Rate, 15 bins')

fig.tight_layout()
plt.show()
```

```python
# (h) Further exploration, del 1: Elite, acceptance rate och korrelationer

# 1. Jämför elit- och icke-elitskolor: medelvärde per variabel
print("Medelvärde per grupp (Elite):")
display(college.groupby('Elite', observed=True).mean(numeric_only=True).T.round(1))

# 2. Median och medel för ett urval nyckelvariabler
key_cols = ['Outstate', 'Expend', 'Grad.Rate', 'S.F.Ratio', 'PhD', 'perc.alumni']
print("Median och medel för nyckelvariabler:")
display(college.groupby('Elite', observed=True)[key_cols].agg(['median', 'mean']).round(1))

# 3. Acceptance rate
college['AcceptRate'] = college['Accept'] / college['Apps']

print("Acceptance rate per Elite:")
display(college.groupby('Elite', observed=True)['AcceptRate'].describe().round(3))
print("Acceptance rate per Private:")
display(college.groupby('Private')['AcceptRate'].describe().round(3))
print("Antal skolor per Private x Elite:")
display(pd.crosstab(college['Private'], college['Elite']))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
college.boxplot(column='AcceptRate', by='Elite', ax=axes[0])
axes[0].set_title('Acceptance rate vs Elite')
college.boxplot(column='AcceptRate', by='Private', ax=axes[1])
axes[1].set_title('Acceptance rate vs Private')
axes[2].scatter(college['AcceptRate'], college['Grad.Rate'], alpha=0.5)
axes[2].set_xlabel('Acceptance rate')
axes[2].set_ylabel('Grad.Rate')
axes[2].set_title('Grad.Rate vs acceptance rate')
fig.suptitle('')
fig.tight_layout()
plt.show()

# 4. Korrelationer med Grad.Rate
corr_grad = college.corr(numeric_only=True)['Grad.Rate'].drop('Grad.Rate').sort_values()
print("Korrelation med Grad.Rate:")
display(corr_grad.round(3))

fig, ax = plt.subplots(figsize=(8, 5))
corr_grad.plot.barh(ax=ax)
ax.set_xlabel('Korrelation med Grad.Rate')
ax.axvline(0, color='black', linewidth=0.8)
fig.tight_layout()
plt.show()

# 5. Felaktiga värden: Grad.Rate över 100 %
print("Skolor med Grad.Rate > 100:")
display(college.loc[college['Grad.Rate'] > 100, ['Private', 'Apps', 'Grad.Rate']])
```

#### Yield: hur eftertraktad är skolan bland de antagna?

```python
# (h) del 2: Yield, dvs andelen antagna som faktiskt börjar
college['Yield'] = college['Enroll'] / college['Accept']

print("Yield per Private och Elite:")
display(college.groupby(['Private', 'Elite'], observed=True)['Yield'].describe().round(3))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
college.boxplot(column='Yield', by='Private', ax=axes[0])
axes[0].set_title('Yield vs Private')
college.boxplot(column='Yield', by='Elite', ax=axes[1])
axes[1].set_title('Yield vs Elite')
for name, grp in college.groupby('Private'):
    axes[2].scatter(grp['AcceptRate'], grp['Yield'], alpha=0.5, label=f'Private = {name}')
axes[2].set_xlabel('Acceptance rate')
axes[2].set_ylabel('Yield')
axes[2].set_title('Yield vs acceptance rate')
axes[2].legend()
fig.suptitle('')
fig.tight_layout()
plt.show()

print("Skolor med lägst acceptance rate och högst yield:")
display(college.sort_values('AcceptRate')[['Private', 'Elite', 'Apps', 'AcceptRate', 'Yield', 'Grad.Rate']].head(10).round(3))
```

#### Kostnad per student i förhållande till tuition

```python
# (h) del 3: Kostnad per student i förhållande till tuition
college['ExpendRatio'] = college['Expend'] / college['Outstate']

print("Expend / Outstate per Private och Elite:")
display(college.groupby(['Private', 'Elite'], observed=True)['ExpendRatio'].describe().round(2))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
college.boxplot(column='ExpendRatio', by='Private', ax=axes[0])
axes[0].axhline(1, color='red', linestyle='--', linewidth=1)
axes[0].set_title('Expend / Outstate vs Private')
for name, grp in college.groupby('Private'):
    axes[1].scatter(grp['Outstate'], grp['Expend'], alpha=0.5, label=f'Private = {name}')
lim = max(college['Outstate'].max(), college['Expend'].max())
axes[1].plot([0, lim], [0, lim], color='red', linestyle='--', linewidth=1, label='Expend = Outstate')
axes[1].set_xlabel('Outstate')
axes[1].set_ylabel('Expend')
axes[1].set_title('Expend vs Outstate')
axes[1].legend()
fig.suptitle('')
fig.tight_layout()
plt.show()

print("Skolor som lägger mest per student i förhållande till tuition:")
display(college.sort_values('ExpendRatio', ascending=False)[['Private', 'Outstate', 'Expend', 'ExpendRatio']].head(5).round(2))
```

#### Skolstorlek

```python
# (h) del 4: Skolstorlek (kvartiler av F.Undergrad)
college['SizeClass'] = pd.qcut(college['F.Undergrad'], 4, labels=['Small', 'Medium', 'Large', 'Very large'])

print("Gränser för storleksklasserna (antal heltidsstudenter):")
display(college.groupby('SizeClass', observed=True)['F.Undergrad'].agg(['min', 'max', 'count']))

print("Nyckelvariabler per storleksklass (median):")
display(college.groupby('SizeClass', observed=True)[['Outstate', 'S.F.Ratio', 'Grad.Rate', 'AcceptRate', 'perc.alumni']].median().round(2))

print("Andel privata per storleksklass:")
display((college.groupby('SizeClass', observed=True)['Private'].apply(lambda s: (s == 'Yes').mean())).round(2))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
for ax, col in zip(axes, ['S.F.Ratio', 'Grad.Rate', 'Outstate']):
    college.boxplot(column=col, by='SizeClass', ax=ax)
    ax.set_title(f'{col} vs skolstorlek')
    ax.set_xlabel('')
fig.suptitle('')
fig.tight_layout()
plt.show()
```

#### Andel deltidsstudenter

```python
# (h) del 5: Andel deltidsstudenter, en proxy för pendlarskola vs campus-skola
college['PartTimeShare'] = college['P.Undergrad'] / (college['F.Undergrad'] + college['P.Undergrad'])
college['PartTimeClass'] = pd.qcut(college['PartTimeShare'], 4, labels=['Q1 (lägst)', 'Q2', 'Q3', 'Q4 (högst)'])

print("Korrelation mellan deltidsandel och övriga variabler:")
display(college.corr(numeric_only=True)['PartTimeShare'].drop('PartTimeShare').sort_values().round(3))

print("Nyckelvariabler per kvartil av deltidsandel (median):")
display(college.groupby('PartTimeClass', observed=True)[['PartTimeShare', 'Grad.Rate', 'perc.alumni', 'Outstate']].median().round(3))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.5))
college.boxplot(column='Grad.Rate', by='PartTimeClass', ax=axes[0])
axes[0].set_title('Grad.Rate vs deltidsandel')
college.boxplot(column='perc.alumni', by='PartTimeClass', ax=axes[1])
axes[1].set_title('perc.alumni vs deltidsandel')
college.boxplot(column='PartTimeShare', by='Private', ax=axes[2])
axes[2].set_title('Deltidsandel vs Private')
for ax in axes:
    ax.set_xlabel('')
fig.suptitle('')
fig.tight_layout()
plt.show()
```

#### Extremvärden och datakvalitet

```python
# (h) del 6: Extremvärden och datakvalitet
print("Variabler i procent med värden över 100:")
for col in ['Grad.Rate', 'PhD', 'Terminal', 'Top10perc', 'Top25perc', 'perc.alumni']:
    bad = college.loc[college[col] > 100, [col]]
    if len(bad):
        print(f"  {col}:")
        display(bad)

print("Fem skolor med lägst Grad.Rate:")
display(college.sort_values('Grad.Rate')[['Private', 'Grad.Rate', 'PartTimeShare', 'Outstate', 'S.F.Ratio']].head(5).round(2))

print("Fem skolor med högst Personal (uppskattade personliga utgifter):")
display(college.sort_values('Personal', ascending=False)[['Private', 'Personal', 'Grad.Rate', 'PartTimeShare']].head(5).round(2))

print("Fem skolor med högst S.F.Ratio:")
display(college.sort_values('S.F.Ratio', ascending=False)[['Private', 'S.F.Ratio', 'Grad.Rate', 'PartTimeShare']].head(5).round(2))
```

#### Korrelationsmatris

```python
import numpy as np
# (h) del 7: Korrelationsmatris över alla numeriska variabler
num_cols = ['Apps', 'Accept', 'Enroll', 'Top10perc', 'Top25perc', 'F.Undergrad', 'P.Undergrad',
            'Outstate', 'Room.Board', 'Books', 'Personal', 'PhD', 'Terminal', 'S.F.Ratio',
            'perc.alumni', 'Expend', 'Grad.Rate']
corr = college[num_cols].corr()

fig, ax = plt.subplots(figsize=(11, 9))
im = ax.imshow(corr, cmap='RdBu_r', vmin=-1, vmax=1)
ax.set_xticks(range(len(num_cols)))
ax.set_yticks(range(len(num_cols)))
ax.set_xticklabels(num_cols, rotation=90)
ax.set_yticklabels(num_cols)
for i in range(len(num_cols)):
    for j in range(len(num_cols)):
        ax.text(j, i, f'{corr.iloc[i, j]:.2f}', ha='center', va='center', fontsize=7,
                color='white' if abs(corr.iloc[i, j]) > 0.6 else 'black')
fig.colorbar(im, ax=ax, shrink=0.8, label='Korrelation')
ax.set_title('Korrelationsmatris, College')
fig.tight_layout()
plt.show()

# De starkaste parvisa korrelationerna (utom diagonalen)
pairs = corr.where(np.triu(np.ones(corr.shape, dtype=bool), k=1)).stack().sort_values(ascending=False)
print("Starkaste positiva korrelationer:")
display(pairs.head(8).round(3))
print("Starkaste negativa korrelationer:")
display(pairs.tail(5).round(3))
```

#### Regression av Grad.Rate

```python
# (h) del 8: Enkel multipel regression av Grad.Rate (förberedelse för kapitel 3)
import numpy as np
import statsmodels.api as sm

# Ta bort det omöjliga värdet 118 % innan regressionen
reg_data = college[college['Grad.Rate'] <= 100].copy()
reg_data['PrivateYes'] = (reg_data['Private'] == 'Yes').astype(int)

X = reg_data[['Outstate', 'perc.alumni', 'S.F.Ratio', 'PartTimeShare', 'Top10perc', 'PrivateYes']]
X = sm.add_constant(X)
y = reg_data['Grad.Rate']

model = sm.OLS(y, X).fit()
print(model.summary())
```
