```python
# Exercise 8
auto = load_data('auto')
import statsmodels.api as sm
X = auto['horsepower']
Y = auto['mpg']
X = sm.add_constant(X)
lin_mod = sm.OLS(Y, X)
lin_reg = lin_mod.fit()
print(lin_reg.summary())
print(f'The predicted miles per gallon with 98 horsepowers: {-0.1578*98 + 39.9359}')
```

```python
from scipy.stats import t


def get_residual_se(y_true, y_pred, df):
    # RSS = Sum of squared residual errors
    RSS = ((y_true - y_pred) ** 2).sum()
    s = (RSS / df) ** 0.5
    return s


def get_pred_se(x, x0, CI=True):
    n = len(x)  # Rätt: len(x) istället för len(auto)
    x_mean = sum(x) / n
    s_xx = sum((x - x_mean) ** 2)

    if CI:
        inside = (1 / n) + ((x0 - x_mean) ** 2) / s_xx 
    else:
        inside = (
            1 + (1 / n) + ((x0 - x_mean) ** 2) / s_xx
        )  

    return inside**0.5


def get_CI(x0, ci_level, p, data, x_col, y_col, model):
    n = len(data)  
    df = n - p
    alpha = 1 - ci_level

    y_0_hat = model.predict([1.0, x0])[0]
    t_crit = float(t.ppf(1 - alpha / 2, df=df))

    s = get_residual_se(data[y_col], model.fittedvalues, df)
    scale_factor = get_pred_se(data[x_col], x0, CI=True)  

    margin = t_crit * s * scale_factor
    return [float(y_0_hat - margin), float(y_0_hat + margin)]


def get_PI(x0, ci_level, p, data, x_col, y_col, model):
    n = len(data)  
    df = n - p
    alpha = 1 - ci_level

    y_0_hat = model.predict([1.0, x0])[0]
    t_crit = float(t.ppf(1 - alpha / 2, df=df))

    s = get_residual_se(data[y_col], model.fittedvalues, df)
    scale_factor = get_pred_se(data[x_col], x0, CI=False)  

    margin = t_crit * s * scale_factor
    return [float(y_0_hat - margin), float(y_0_hat + margin)]


# Körning
x0 = 98
ci = get_CI(
    x0,
    ci_level=0.95,
    p=2,
    data=auto,
    x_col='horsepower',
    y_col='mpg',
    model=lin_reg,
)
pi = get_PI(
    x0,
    ci_level=0.95,
    p=2,
    data=auto,
    x_col='horsepower',
    y_col='mpg',
    model=lin_reg,
)

print(f'95% CI for x0={x0}: {ci}')
print(f'95% PI for x0={x0}: {pi}')
```

```python
def abline(ax, b, m, *args, **kwargs):
    "Add a line with slope m and intercept b to ax"
    xlim = ax.get_xlim()
    ylim = [m * xlim[0] + b, m * xlim[1] + b]
    ax.plot(xlim, ylim, *args, **kwargs)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(auto['horsepower'], auto['mpg'], alpha=0.5)
abline(ax, lin_reg.params['const'], lin_reg.params['horsepower'], 'r--', linewidth=2)
ax.set_xlabel('horsepower'); ax.set_ylabel('mpg')
plt.show()
```

```python
# Diagnostic plots
import numpy as np, scipy.stats as stats
from statsmodels.nonparametric.smoothers_lowess import lowess

def diagnostic_plots(results, n_label=3):
    """The four plots of R's plot(lm) for a fitted statsmodels OLS result."""
    infl = results.get_influence()
    fitted, resid = results.fittedvalues, results.resid
    std_resid = infl.resid_studentized_internal
    leverage, cooks = infl.hat_matrix_diag, infl.cooks_distance[0]
    p, n = results.df_model + 1, len(resid)
    idx = results.model.data.row_labels
    fig, axes = plt.subplots(2, 2, figsize=(12, 9))

    # 1. Residuals vs fitted: curvature = wrong functional form
    ax = axes[0, 0]
    ax.scatter(fitted, resid, alpha=0.5)
    s = lowess(resid, fitted, frac=0.6); ax.plot(s[:, 0], s[:, 1], 'r-')
    ax.axhline(0, color='grey', ls='--')
    ax.set(xlabel='Fitted values', ylabel='Resivs Fitted')
    for i in np.argsort(np.abs(resid.values))[-n_label:]:
        ax.annotate(idx[i], (fitted.iloc[i], resid.iloc[i]), fontsize=8)

    # 2. Normal Q-Q of standardized residuals
    ax = axes[0, 1]
    (osm, osr), (slope, intercept, _) = stats.probplot(std_resid, dist='norm')
    ax.scatter(osm, osr, alpha=0.5); ax.plot(osm, slope*osm + intercept, 'r-')
    ax.set(xlabel='Theoretical quantiles', ylabel='Standardized residuals', title='Normal Q-Q')

    # 3. Scale-location: trend = non-constant v
    ax = axes[1, 0]
    sqrt_abs = np.sqrt(np.abs(std_resid))
    ax.scatter(fitted, sqrt_abs, alpha=0.5)
    s = lowess(sqrt_abs, fitted, frac=0.6); ax.plot(s[:, 0], s[:, 1], 'r-')
    ax.set(xlabel='Fitted values', ylabel=r'$\sqrt{|\mathrm{std.\ residuals}|}$', title='Scale-Location')

    # 4. Residuals vs leverage with Cook's distance contours
    ax = axes[1, 1]
    ax.scatter(leverage, std_resid, alpha=0.5)
    ax.axhline(0, color='grey', ls='--')
    ax.axvline(2*p/n, color='grey', ls=':', label='2p/n')
    lev = np.linspace(leverage.min(), leverage.max(), 100)
    for d in (0.5, 1.0):
        for sign in (1, -1):
            ax.plot(lev, sign*np.sqrt(d*p*(1 - lev)/lev), 'r--', lw=0.8)
    ax.set_ylim(std_resid.min() - 0.5, std_resid.max() + 0.5)
    ax.set(xlabel='Leverage', ylabel='Standardized residuals', title="Residuals vs Leverage")
    for i in np.argsort(cooks)[-n_label:]:
        ax.annotate(idx[i], (leverage[i], std_resid[i]), fontsize=8)
    ax.legend()
    fig.tight_layout()
    return fig

diagnostic_plots(results)
```
