```python
import matplotlib.pyplot as plt
import math
import numpy as np
import random as random

n = 100_000
probs = {}
for i in range(1, n+1):
    p = 1 - (1 - 1 / i) ** i
    probs[i] = p
fig, ax = plt.subplots()

ax.plot(list(probs.keys()), list(probs.values()), label = "Probabilities")
ax.axhline(y=1 - (math.e)**(-1), color="gray", linestyle=":", label="y = 1 - 1/e")   
ax.set_xscale("log")
ax.set(xlabel="n (log scale)", ylabel="P(j-th observation in bootstrap sample)")
ax.legend()
plt.show()
```
