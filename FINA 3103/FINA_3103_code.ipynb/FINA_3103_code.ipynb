```python
E_A = 0.1
E_B = 0.3
rf = 0.05
std_A = 0.2
std_B = 0.6
rho_AB = -0.2

w_A = ((E_A - rf)*std_B**2 - (E_B - rf)*rho_AB*std_A*std_B)/((E_A - rf)*std_B**2 + (E_B - rf)*std_A**2 - (E_A - rf + E_B - rf)*rho_AB*std_A*std_B)
w_B = 1 - w_A
# Expected return and standard deviation of the portfolio
E_P = w_A * E_A + w_B * E_B
std_P = (w_A**2*std_A ** 2 + w_B**2*std_B ** 2 + 2*w_A*w_B*rho_AB*std_B*std_A) ** (1/2)
SR = (E_P - rf)/std_P
print(f'The sharpe ratio of the optimal risky portfolio is {SR}')
print(f'The expected value and the volaility of optimal risky portfolio is {E_P} and {std_P}')
A = 5
y = (E_P - rf) / (A * std_P**2)
w_A_C = y * w_A
w_B_C = y * w_B
w_rf_C = y
print(f'The final weights of the complete portfolio for an investor with risk aversion A=5 is\nA:{w_A_C}\nB:{w_B_C}\nrf:{w_rf_C}')
# For the global minimum variance
cov_AB = std_A*std_B*rho_AB
w_A = (std_B ** 2 - cov_AB)/(std_B ** 2 + std_A ** 2 - 2 * cov_AB)
w_B = 1 - w_A
std_gmv = ((w_A*std_A) ** 2 + (w_B*std_B) ** 2 + 2*w_A*w_B*cov_AB)
print(f'The standard deviation of the gmv portfolio is {std_gmv}')
```
